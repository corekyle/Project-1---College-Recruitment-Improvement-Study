This notebook reads the fall 2024 entry-level computing majors survey dataset, performs exploratory data analysis using built in pandas properties and methods, and performs systematic data cleaning (column renaming, feature selection, category standardization, handling missing values), and exports the cleaned dataset to a new CSV file.

```python
# Import required libraries
import pandas as pd
import numpy as np

# Load the assigned dataset using a relative path
file_path = 'Majors Survey Results - Fall 2024.csv'
df = pd.read_csv(file_path)

# Display the first few rows of the raw dataset
df.head(3)
```

## 2) Data Cleaning Operations
### A: Renaming Columns
Convert all column names to lowercase, strip leading/trailing whitespace, and replace spaces, question marks, parenthesis, and slashes with underscores.

```python
# Create a working copy for data cleaning
df_clean = df.copy()

# Function to standardize column names
def clean_col_name(col):
    col = col.lower().strip()
    col = col.replace('?', '').replace('(', '').replace(')', '').replace('/', '_')
    col = col.replace('  ', ' ').replace(' ', '_').replace('-', '_')
    return col

# Apply renaming to all columns
df_clean.columns = [clean_col_name(c) for c in df_clean.columns]

# Display the first 15 cleaned column names
df_clean.columns[:15]
```

### B: Dropping Irrelevant Columns
Drop unnecessary metadata fields (timestamps and open-ended text explanation columns), that are not required for quantative analysis questions. Preserves all core demographic, motivation, and outreach columns.

```python
# Identify columns to drop (timestamps and open-ended text explanations)
cols_to_drop = [c for c in df_clean.columns if 'timestamp' in c or 'please_explain' in c]

# Drop the columns
df_clean = df_clean.drop(columns=cols_to_drop, errors='ignore')

# Display remaining shape
df_clean.shape
```

### C: Categorical Values Standardization
Standardize categorical values for race and major by removing reduntant text strings from responses.

```python
# Check unique uncleaned major values before transformation
df_clean['what_degree_program_are_you_currently_enrolled_in'].unique()[:15]

# Define mapping function to condense major/degree programs
def condense_major(val):
    if pd.isna(val):
        return 'Undecided/Other'
    val_lower = str(val).lower()
    if 'computer science' in val_lower:
        return 'Computer Science'
    elif 'cyber' in val_lower:
        return 'Cybersecurity'
    elif 'information technology' in val_lower or 'cis' in val_lower:
        return 'Information Technology / CIS'
    elif 'engineering' in val_lower:
        return 'Engineering & Technology'
    elif 'business' in val_lower:
        return 'Business Administration'
    elif 'challenger' in val_lower or 'sharetime' in val_lower:
        return 'High School ShareTime / Challenger'
    else:
        return 'Other / General Studies'

# Apply the condensation function to create a clean major column
df_clean['degree_program_cleaned'] = df_clean['what_degree_program_are_you_currently_enrolled_in'].apply(condense_major)

# Verify the cleaned category distribution
df_clean['degree_program_cleaned'].value_counts()
```

### D: Handle Missing Values
Inspect missing values ('NaN'). Inspect missing counts and apply categorical labels to demographic fields where values were left blank.

```python
# Check top missing value counts across columns
df_clean.isnull().sum().sort_values(ascending=False).head(10)

# Impute missing demographic fields with a clear 'Not Specified' label
df_clean['gender'] = df_clean['gender'].fillna('Not Specified')
df_clean['race_ethnicity'] = df_clean['race_ethnicity'].fillna('Not Specified')
df_clean['age'] = df_clean['age'].fillna('Not Specified')

# Verify null counts for key demographics after imputation
df_clean[['gender', 'race_ethnicity', 'age']].isnull().sum()
```

## 3) Export Cleaned Dataset
Verify the final shape of the cleaned dataset and save it as a new CSV file.

```python
# Verify final dataset shape
df_clean.shape

# Save cleaned dataset to a new CSV file
output_filename = 'Majors Survey Results - Fall 2024 - Cleaned.csv'
df_clean.to_csv(output_filename, index=False)

# Display preview of the finalized cleaned dataframe
df_clean.head(2)
```
